```python
import pandas as pd
import numpy as np

# Load CSVs
files = ["bert_predictions.csv"]
dfs = [pd.read_csv(file) for file in files]

print(dfs[0].columns)
# Combine relevant columns into a single DataFrame
columns = [
    'Predicted label [bert_200_labeled_data_fine_tune_abstract]',
       'Level of Confidence [bert_200_labeled_data_fine_tune_abstract]',
       'Predicted label [bert_200_labeled_data_fine_tune_title]',
       'Level of Confidence [bert_200_labeled_data_fine_tune_title]',
       'Predicted label [bert_200_labeled_data_fine_tune_keywords]',
       'Level of Confidence [bert_200_labeled_data_fine_tune_keywords]',
       'Predicted label [bert_200_labeled_data_fine_tune_all]',
       'Level of Confidence [bert_200_labeled_data_fine_tune_all]'
]

# Extract predictions and confidence
predictions = pd.concat([df[columns] for df in dfs], axis=1)

# Apply weighted voting
def weighted_vote(row):
    # Collect predictions and confidences
    labels = [
        row['Predicted label [bert_200_labeled_data_fine_tune_title]'],
        row['Predicted label [bert_200_labeled_data_fine_tune_all]'],
        row['Predicted label [bert_200_labeled_data_fine_tune_keywords]'],
        row['Predicted label [bert_200_labeled_data_fine_tune_abstract]']
    ]
    confidences = [
        row['Level of Confidence [bert_200_labeled_data_fine_tune_title]'],
        row['Level of Confidence [bert_200_labeled_data_fine_tune_all]'],
        row['Level of Confidence [bert_200_labeled_data_fine_tune_keywords]'],
        row['Level of Confidence [bert_200_labeled_data_fine_tune_abstract]']
    ]
    
    # Calculate weighted scores for each class
    classes = [0, 1]
    weighted_scores = {c: 0 for c in classes}
    for label, confidence in zip(labels, confidences):
        weighted_scores[label] += confidence

    # Choose the class with the highest weighted score
    final_label = max(weighted_scores, key=weighted_scores.get)
    final_confidence = weighted_scores[final_label] / sum(confidences)
    
    return final_label, final_confidence

# Apply to each row
results = predictions.apply(lambda row: weighted_vote(row), axis=1)
predicted_labels, confidences = zip(*results)

# Add to DataFrame
dfs[0]['Final Predicted Label'] = predicted_labels
dfs[0]['Final Confidence'] = confidences

# Save final results
dfs[0].to_csv("final_predictions.csv", index=False)
print("Final predictions saved to final_predictions.csv")
```
