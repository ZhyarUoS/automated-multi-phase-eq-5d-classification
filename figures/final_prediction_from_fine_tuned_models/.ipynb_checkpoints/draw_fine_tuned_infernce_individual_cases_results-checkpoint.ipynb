```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the Excel file
file_path = 'merged_predictions.xlsx'  # Replace with the actual file path
data = pd.read_excel(file_path)

# Define the color mapping for binary classes (0 and 1) with confidence levels
color_mapping = {
    1: {  # Class 1
        "Not sure (< 0.50)": "#d73027",
        "Uncertainty (0.50-0.60)": "#fc8d59",
        "Low confidence (0.60-0.70)": "#fee08b",
        "High confidence (0.70-0.85)": "#d9ef8b",
        "Confident (> 0.85)": "#1a9850"
    },
    0: {  # Class 0
        "Not sure (< 0.50)": "#762a83",
        "Uncertainty (0.50-0.60)": "#af8dc3",
        "Low confidence (0.60-0.70)": "#e7d4e8",
        "High confidence (0.70-0.85)": "#7fbf7b",
        "Confident (> 0.85)": "#1b7837"
    }
}

# Function to determine the confidence level category
def categorize_confidence(conf):
    if conf < 0.50:
        return "Not sure (< 0.50)"
    elif 0.50 <= conf < 0.60:
        return "Uncertainty (0.50-0.60)"
    elif 0.60 <= conf < 0.70:
        return "Low confidence (0.60-0.70)"
    elif 0.70 <= conf < 0.85:
        return "High confidence (0.70-0.85)"
    else:
        return "Confident (> 0.85)"

# Create an output directory for the figures
output_dir = 'prediction_pie_charts_individual'
os.makedirs(output_dir, exist_ok=True)

# Iterate through each study to generate a pie chart
for index, row in data.iterrows():
    # Extract study information
    study_id = row['N']
    doi = row['Doi']
    
    # Model predictions and confidence scores
    models = ['BERT_A', 'BERT_T' ,'BERT_K', 'BERT_C', 'BioBERT_A', 'BioBERT_T', 'BioBERT_K', 'BioBERT_C', 'SciBERT_A',
             'SciBERT_T', 'SciBERT_K', 'SciBERT_C'
             ]
    predictions = [
        row['bert_pred_abstract'],
        row['bert_pred_title'],
        row['bert_pred_keywords'],
        row['bert_pred_combined'],
        row['biobert_pred_abstract'],
        row['biobert_pred_title'],
        row['biobert_pred_keywords'],
        row['biobert_pred_combined'],
        row['scibert_pred_abstract'],
        row['scibert_pred_title'],
        row['scibert_pred_keywords'],
        row['scibert_pred_combined']
    ]
    confidences = [
       row['bert_confi_abstract'],
        row['bert_confi_title'],
        row['bert_confi_keywords'],
        row['bert_confi_combined'],
        row['biobert_confi_abstract'],
        row['biobert_confi_title'],
        row['biobert_confi_keywords'],
        row['biobert_confi_combined'],
        row['scibert_confi_abstract'],
        row['scibert_confi_title'],
        row['scibert_confi_keywords'],
        row['scibert_confi_combined']
    ]
    
    # Determine colors for each model's prediction
    categories = [categorize_confidence(conf) for conf in confidences]
    colors = [
        color_mapping[pred][category]
        for pred, category in zip(predictions, categories)
    ]
    
    # Prepare pie chart labels
    labels = [
        f"{model}\nClass: {pred}\nConf: {conf:.1f}%"
        for model, pred, conf in zip(models, predictions, confidences)
    ]
    
    # Plot the pie chart
    plt.figure(figsize=(8, 5))  # Wider figure for compact layout
    plt.pie(
        confidences, 
        labels=labels, 
        colors=colors, 
        startangle=140
    )
    plt.title(f'Study {study_id}', fontsize=12)  # Smaller title font
    
    # Add a legend explaining the color coding
    legend_handles = []
    for cls, mappings in color_mapping.items():
        for level, color in mappings.items():
            legend_handles.append(
                plt.Line2D([0], [0], color=color, lw=6, label=f"Class {cls}: {level}")
            )
    plt.legend(
        handles=legend_handles,
        title="Confidence Levels",
        loc='center left',
        bbox_to_anchor=(1.05, 0.5),  # Place the legend to the right of the pie chart
        fontsize=8,
        frameon=False
    )
    
    # Add study details
    plt.annotate(
        f'DOI: {doi}', 
        xy=(0.5, -0.2),  # Adjusted position below the pie chart
        xycoords='axes fraction', 
        fontsize=8, 
        ha='center'
    )
    
    # Save the figure
    output_path = os.path.join(output_dir, f'study_{index + 1}.png')
    plt.savefig(output_path, bbox_inches='tight', dpi=150)  # Reduced DPI for smaller file size
    plt.close()
```
