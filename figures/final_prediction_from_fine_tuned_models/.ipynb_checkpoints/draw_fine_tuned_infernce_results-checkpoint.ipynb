```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load the Excel file
file_path = 'merged_predictions.xlsx'  # Replace with the actual file path
data = pd.read_excel(file_path)

# Define the color mapping for binary classes (0 and 1) with confidence levels
color_mapping = {
    0: {  # Class 0
        "Not sure (< 50)": "#d73027",
        "Uncertainty (50-60)": "#fc8d59",
        "Low confidence (60-70)": "#fee08b",
        "High confidence (70-85)": "#d9ef8b",
        "Confident (> 85)": "#1a9850"
    },
    1: {  # Class 1
        "Not sure (< 50)": "#762a83",
        "Uncertainty (50-60)": "#af8dc3",
        "Low confidence (60-70)": "#e7d4e8",
        "High confidence (70-85)": "#7fbf7b",
        "Confident (> 85)": "#1b7837"
    }
}

# Function to determine the confidence level category
def categorize_confidence(conf):
    if conf < 50:
        return "Not sure (< 50)"
    elif 50 <= conf < 60:
        return "Uncertainty (50-60)"
    elif 60 <= conf < 70:
        return "Low confidence (60-70)"
    elif 70 <= conf < 85:
        return "High confidence (70-85)"
    else:
        return "Confident (> 85)"

# Create an output directory for the figures
output_dir = 'prediction_pie_charts'
os.makedirs(output_dir, exist_ok=True)

# Iterate through each study to generate a pie chart
for index, row in data.iterrows():
    # Extract study information
    study_id = row['N']
    doi = row['Doi']
    
    # Model predictions and confidence scores
    models = ['BERT', 'BioBERT', 'SciBERT']
    predictions = [
        row['bert_Final Predicted Label'],
        row['biobert_Final Predicted Label'],
        row['scibert_Final Predicted Label']
    ]
    confidences = [
        row['bert_Final Confidence'],
        row['biobert_Final Confidence'],
        row['scibert_Final Confidence']
    ]
    
    # Determine colors for each model's prediction
    categories = [categorize_confidence(conf) for conf in confidences]
    colors = [
        color_mapping[pred][category]
        for pred, category in zip(predictions, categories)
    ]
    
    # Prepare pie chart labels (confidence level inside slices)
    labels = [
        f"{model}\nConf: {conf:.1f}%"
        for model, conf in zip(models, confidences)
    ]
    
    # Plot the pie chart
    plt.figure(figsize=(10, 8))
    wedges, texts = plt.pie(
        confidences, 
        labels=None,  # Remove default labels
        colors=colors, 
        startangle=140
    )
    
    # Add confidence levels inside slices
    for i, wedge in enumerate(wedges):
        plt.text(
            wedge.theta2 - (wedge.theta2 - wedge.theta1) / 2,
            0.5,
            labels[i],
            ha='center',
            fontsize=10,
            rotation=(wedge.theta2 - wedge.theta1) / 2,
            color='white',
            weight='bold',
        )
    
    # Add a title
    plt.title('Prediction Confidence by Model', fontsize=16)
    
    # Add a legend explaining the color coding
    legend_labels = [
        f"Class {cls}: {level}" 
        for cls, mapping in color_mapping.items()
        for level in mapping.keys()
    ]
    legend_colors = [
        color for cls, mapping in color_mapping.items() for color in mapping.values()
    ]
    plt.legend(
        handles=[
            plt.Line2D([0], [0], color=color, lw=4) 
            for color in legend_colors
        ],
        labels=legend_labels,
        title="Confidence Levels",
        loc='upper right',
        bbox_to_anchor=(1.3, 1)
    )
    
    # Add study details
    plt.annotate(f'Study: {study_id}', xy=(0, -1.2), xycoords='axes fraction', fontsize=10, ha='center')
    plt.annotate(f'DOI: {doi}', xy=(0, -1.4), xycoords='axes fraction', fontsize=10, ha='center')
    
    # Save the figure
    output_path = os.path.join(output_dir, f'study_{index + 1}.png')
    plt.savefig(output_path, bbox_inches='tight')
    plt.close()
```
