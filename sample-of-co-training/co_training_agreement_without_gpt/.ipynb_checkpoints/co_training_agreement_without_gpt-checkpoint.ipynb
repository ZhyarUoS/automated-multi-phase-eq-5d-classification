```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV files
scibert_bert = pd.read_csv('co_training_agreement_without_gpt/scibert_bert.csv')
bert_biobert = pd.read_csv('co_training_agreement_without_gpt/bert_biobert.csv')
scibert_biobert = pd.read_csv('co_training_agreement_without_gpt/scibert_biobert.csv')

# Extract relevant columns and rename predictions
df1 = scibert_bert[['N', 'final_prediction']].rename(columns={'final_prediction': 'pred1'})
df2 = bert_biobert[['N', 'final_prediction']].rename(columns={'final_prediction': 'pred2'})
df3 = scibert_biobert[['N', 'final_prediction']].rename(columns={'final_prediction': 'pred3'})

# Merge on study ID 'N'
merged = df1.merge(df2, on='N').merge(df3, on='N')

# Define agreement category
def agreement(row):
    preds = [row['pred1'], row['pred2'], row['pred3']]
    unique_preds = set(preds)
    if len(unique_preds) == 1:
        return 'Full agreement'
    elif len(unique_preds) == 2:
        return 'At least two agree'
    else:
        return 'No agreement'

merged['agreement'] = merged.apply(agreement, axis=1)

# Summary counts and percentages
summary = merged['agreement'].value_counts().reindex(['Full agreement', 'Two models agreed', 'No agreement'], fill_value=0)
total = summary.sum()
percentages = (summary / total * 100).round(2)

# Create summary DataFrame
summary_df = pd.DataFrame({
    'Category': summary.index,
    'Number of Records': summary.values,
    'Percentage': percentages.values
})

# Save summary CSV
summary_df.to_csv('agreement_summary.csv', index=False)

# Plot bar chart with numbers and percentages on bars
plt.figure(figsize=(8,6))
bars = plt.bar(summary_df['Category'][:2], summary_df['Number of Records'][:2], color=['green', 'orange'])

plt.title('Model Agreement Summary')
plt.ylabel('Number of Studies')

# Add text annotations (number and percentage) on bars
for bar, num, pct in zip(bars, summary_df['Number of Records'][:2], summary_df['Percentage'][:2]):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + total*0.01, f'{num}\n({pct}%)', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.savefig('agreement_summary_bar.png')
plt.show()
```
