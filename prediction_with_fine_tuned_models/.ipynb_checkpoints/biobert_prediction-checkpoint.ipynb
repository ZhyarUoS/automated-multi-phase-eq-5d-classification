```python
import os
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from tqdm import tqdm
import re
import numpy as np

# File paths
data_file = "../data/random_200_records.csv"
models_dir = "bert"
output_file = "bert_predictions.csv"

# Clean text function
def clean_text(text):
    if isinstance(text, (str, np.ndarray)): 
        if isinstance(text, str):
            text = text.lower()
            text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
            return text
        elif isinstance(text, np.ndarray):
            return np.array([clean_text(t) for t in text])
    elif pd.isnull(text): 
        return ""
    else:
        try:
            text = str(text)
            text = text.lower()
            text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
            return text
        except Exception as e:
            print(f"Error cleaning text: {e}, Text Type: {type(text)}, Text Value: {text}")
            return "" 

# Load data
data = pd.read_csv(data_file)

# Clean text columns
for col in ['Title', 'Abstract', 'Keywords']:
    data[col] = data[col].apply(clean_text)

# Combine cleaned text for tokenization
data['combined_text'] = data['Title'] + " [SEP] " + data['Abstract'] + " [SEP] " + data['Keywords']

# Initialize tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenization function
def tokenize_data(texts, max_length=128):
    return tokenizer(
        texts,
        max_length=max_length,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )

# Prepare dataset for inference
tokenized_data = tokenize_data(data['combined_text'].tolist())
input_ids = tokenized_data['input_ids']
attention_mask = tokenized_data['attention_mask']

# Create a DataFrame to store predictions
results_df = data.copy()

# Iterate over all model folders
model_folders = [
    f for f in os.listdir(models_dir)
    if os.path.isdir(os.path.join(models_dir, f)) and not f.startswith('.')  # Exclude hidden folders
    and os.path.exists(os.path.join(models_dir, f, "model.safetensors"))  # Ensure required files exist
    and os.path.exists(os.path.join(models_dir, f, "config.json"))
]


for folder in tqdm(model_folders, desc="Processing Models"):
    folder_path = os.path.join(models_dir, folder)
    model_path = os.path.join(folder_path, "model.safetensors")
    config_path = os.path.join(folder_path, "config.json")
    
    # Load model
    model = AutoModelForSequenceClassification.from_pretrained(folder_path)
    model.eval()
    model.to("cuda" if torch.cuda.is_available() else "cpu")
    
    # Inference
    predictions = []
    confidence_scores = []
    with torch.no_grad():
        for i in range(0, len(data), 32):  # Batch inference
            batch_input_ids = input_ids[i:i+32].to("cuda" if torch.cuda.is_available() else "cpu")
            batch_attention_mask = attention_mask[i:i+32].to("cuda" if torch.cuda.is_available() else "cpu")
            
            outputs = model(batch_input_ids, attention_mask=batch_attention_mask)
            logits = outputs.logits
            probs = torch.softmax(logits, dim=-1).cpu()
            
            batch_predictions = torch.argmax(probs, dim=1).tolist()
            batch_confidence = probs.max(dim=1).values.tolist()
            
            predictions.extend(batch_predictions)
            confidence_scores.extend(batch_confidence)
    
    # Add predictions to results_df
    results_df[f'Predicted label [{folder}]'] = predictions
    results_df[f'Level of Confidence [{folder}]'] = confidence_scores

# Save results to CSV
results_df.to_csv(output_file, index=False)
print(f"Predictions saved to {output_file}")
```
